AIM: Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas

Step 1: Create DataFrame


```python
# Import pandas library and create a student dataset.
# The dataset contains ID, Name, Age, City, and Marks.
import pandas as pd

data = {
    "ID": [101,102,103,104,104,105,106,107],
    "Name": ["Rahul","Priya","Amit",None,"Riya","Ajay","Naresh","jenil"],
    "Age": [20,21,None,20,20,22,23,28],
    "city": ["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahemdabad"],
    "Marks": [85,None,78,88,88,81,90,93]
}

df = pd.DataFrame(data)
df
```

Step 2: Inspect the dataset

```python
# Display the first 5 rows of the dataset.
df.head()
```

```python
# Display the last 5 rows and columns.
df.tail()
```

```python
# Display detailed information about the dataset.
df.info()
```

```python

# Display statistical information of numerical columns.
df.describe()
```

```python
# Display the dimension of the dataset.
# Output: (number of rows, number of columns)
df.shape
```

```python
# Display the data type of each column.
df.dtypes
```

```python
# Display all columns of the DataFrame.
df.columns

```

 3: Check missing values

```python
# Check for missing values in the dataset.
df.isnull()
```

```python
# Count missing values in each column.
df.isnull().sum()
```

```python
# Count total missing values in the whole dataset.
df.isnull().sum().sum()
```

 4: Fill missing values

```python
# Replace the missing value in the Marks column with the mean.
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
# Replace the missing value in the Age column with the mean.
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
# Replace the missing value in the Name column with "unknown".
df["Name"] = df["Name"].fillna("unknown")
```

```python
# Display the updated DataFrame.
df
```

5: Detect duplicate records

```python
# Check whether each row is duplicated.
df.duplicated()
```

```python
# Count the total number of duplicate rows.
df.duplicated().sum()

```

6: Remove duplicate records

```python
# Remove duplicate rows from the DataFrame.
df.drop_duplicates(inplace=True)

# Display the updated DataFrame.
df
```

7: Rename columns

```python
# Rename the Marks column to Score.
df.rename(columns={"Marks":"Score"}, inplace=True)

# Display the updated DataFrame.
df
```

 8: Change data type

```python
# Convert the Age column into integer data type.
df["Age"] = df["Age"].astype(int)

# Display the updated DataFrame.
df
```

9: Calculate final score

```python
# Add 5 marks to the Score column and create a new column.
df["finnal score"] = df["Score"] + 5

# Display the updated DataFrame.
df
```

 10: Convert names into uppercase

```python
# Convert all names into uppercase letters.
df["Name"] = df["Name"].str.upper()

# Display the updated DataFrame.
df
```

 11: Convert city names into lowercase

```python
# Convert all city names into lowercase letters.
df["city"] = df["city"].str.lower()

# Display the updated DataFrame.
df
```

12: Remove extra spaces

```python
# Remove unwanted spaces from city names.
df["city"] = df["city"].str.strip()

# Display the updated DataFrame.
df
```

13: Sort dataset

```python
# Sort the dataset according to final score.
# ascending=False sorts the data from highest to lowest.
df = df.sort_values("finnal score", ascending=False)

# Display the sorted DataFrame.
df
```

14: Save clean dataset

```python
# Save the cleaned DataFrame as a CSV file.
# index=False prevents row numbers from being saved.
df.to_csv("cleaned_students.csv", index=False)

print("Dataset Saved Successfully.")
```
