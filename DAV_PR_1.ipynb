AIM: Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON Files using Pandas.

 Step 1: Import pandas library

```python
# Pandas is used to create, read, modify, and analyze data in table format.
import pandas as pd
```


 Step 2: Create student data

```python
# Here we create a dictionary containing student ID, name, age, city, and marks.
data = {
    "ID": [101,102,103,104,105,106,107],
    "Name":["Rahul","Priya","Amit","Riya","Karan","krisha","harry"],
    "Age":[20,21,19,20,22,24,28],
    "City":["surat","navsari","valsad","bharuch","surat","gandhinagar","ahemdabad"],
    "Marks":[85,92,78,88,81,84,90]
}
```

Step 3: Create a DataFrame


```python
# pd.DataFrame() converts the dictionary into a table with rows and columns.
df = pd.DataFrame(data)
df
```

Step 4: Save the DataFrame as a CSV file

```python
# to_csv() saves the data in CSV format.
# index=False prevents pandas from saving row numbers in the file.
df.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

Step 5: Save the DataFrame as an Excel file

```python
# to_excel() saves the data in Microsoft Excel format.
# index=False prevents row numbers from being saved.
df.to_excel("students.xlsx", index=False)
print("Excel file created successfully.")

```

Step 6: Save the DataFrame as a JSON file

```python
# to_json() saves the data in JSON format.
# orient="records" stores each row as a separate JSON record.
df.to_json("students.json", orient="records")
print("JSON file created successfully.")
```

Step 7: Read the CSV file

```python
# read_csv() loads the saved CSV file into a new DataFrame.
df_csv = pd.read_csv("students.csv")
df_csv
```

Step 8: Read the Excel file


```python
# read_excel() loads the saved Excel file into a DataFrame.
df_excel = pd.read_excel("students.xlsx")
df_excel
```

Step 9: Read the JSON file

```python
# read_json() loads the saved JSON file into a DataFrame.
df_json = pd.read_json("students.json")
df_json
```
